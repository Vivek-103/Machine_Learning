## Support Vector Regression

### Importing the Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

### Importing the Dataset

```python
dataset = pd.read_csv('Position_Salaries.csv')
x = dataset.iloc[:,1:-1].values
y = dataset.iloc[:,-1].values
```

```python
print(x)
```

```python
print(y)
```

```python
y = y.reshape(len(y),1)
```

```python
print(y)
```

### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
sc_y = StandardScaler()
x = sc_x.fit_transform(x)
y = sc_y.fit_transform(y)
```

```python
print(x)
```

```python
print(y)
```

### Training the SVR model on the whole dataset

```python
from sklearn.svm import SVR
regressor = SVR(kernel = 'rbf')
regressor.fit(x,y)
```

### Predict the New Result

```python
sc_y.inverse_transform(regressor.predict(sc_x.transform([[6.5]])).reshape(-1,1))
```

### Visualize the SVR result

```python
plt.scatter(sc_x.inverse_transform(x),sc_y.inverse_transform(y),color='red')
plt.plot(sc_x.inverse_transform(x),sc_y.inverse_transform(regressor.predict(x).reshape(-1,1)),color='blue')
plt.title('SVR RESULT')
plt.xlabel('Position Level')
plt.ylabel('Salaray')
plt.show()
```

### SVR (in HIGHER RESULOTION)

```python
x_grid = np.arange(np.min(sc_x.inverse_transform(x)), np.max(sc_x.inverse_transform(x)), 0.1)
x_grid = x_grid.reshape((len(x_grid), 1))

plt.scatter(sc_x.inverse_transform(x), sc_y.inverse_transform(y), color='red')
plt.plot(
    x_grid,
    sc_y.inverse_transform(regressor.predict(sc_x.transform(x_grid)).reshape(-1, 1)),
    color='blue'
)
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```
